# Week 1 - Pizza Chatbot

## Assignment: Pizza Chatbot Development

Create a chatbot that takes a standard 3-step pizza order using a state variable.

Step 1: Ask the user what size pizza they want (Small, Medium, Large).

Step 2: Ask what single topping they want (Pepperoni, Mushrooms, Cheese).

Step 3: Confirm the order by repeating their choices back to them (e.g., "Awesome, I have recorded a Medium Pepperoni pizza for you! Goodbye!").

Bonus constraint: If the user types "bye" at any point, the bot must stop immediately.

```python
def pizza_bot():
    print("Welcome to Pizza Bot!")

    prices = {"small": "$8", "medium": "$12", "large": "$16"}
    valid_toppings = ["pepperoni", "mushrooms", "cheese"]

    show_menu = input("Can we show you our menu? (yes/no): ")
    if show_menu.lower() == "bye":
        print("Goodbye!")
        return

    if show_menu.lower() == "yes":
        print("Sizes: Small ($8), Medium ($12), Large ($16)")
        print("Toppings: Pepperoni, Mushrooms, Cheese")

    while True:
        size = input("What would you like to have from our menu? Choose a pizza size (Small, Medium, Large): ")
        if size.lower() == "bye":
            print("Goodbye!")
            return
        if size.lower() in prices:
            break
        print("Sorry, that's not a size we have. Please choose Small, Medium, or Large.")

    while True:
        topping = input("Choose a topping (Pepperoni, Mushrooms, Cheese): ")
        if topping.lower() == "bye":
            print("Goodbye!")
            return
        if topping.lower() in valid_toppings:
            break
        print("Sorry, that's not a topping we have. Please choose Pepperoni, Mushrooms, or Cheese.")

    print(f"Awesome, I have recorded a {size.capitalize()} {topping.capitalize()} pizza for you! Goodbye!")

```

```python
pizza_bot()
```
